```python
# chat_models 
from langchain_google_genai import ChatGoogleGenerativeAI

llm = ChatGoogleGenerativeAI(
    model = "gemini-flash-latest",
    max_tokens = 200
)

from langchain_core.messages import HumanMessage, SystemMessage

messages = [
    SystemMessage("You are a Doctor who has knowledge in AI and explains every AI concept from the perspective of biology."),
    HumanMessage("What is RAG?")
]


for token in llm.stream(messages):
    print(token)
```

```python
# embedding models
from langchain_openai import OpenAIEmbeddings

embeddings = OpenAIEmbeddings(model="text-embedding-3-small")
# can't actually run without an Open AI API key.
# Convert a sentence to a vector
vector = embeddings.embed_query("What is machine learning?")
print(len(vector))    # 1536 dimensions
print(vector[:5])     # [-0.023, 0.041, -0.018, 0.029, -0.011, ...]
```

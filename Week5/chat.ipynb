```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI as ChatG
llm_gemini = ChatG(model = "gemini-flash-latest")
```

```python
# test call to gemini model 
llm_gemini.invoke("What is Langchain?").content
```

```python
system_prompt = "You have to behave like a movie star who is gay and is expert in teaching AI to beginners."

user_prompt = "What is Langchain?"

from langchain_core.messages import HumanMessage, SystemMessage
import textwrap

messages = [
    SystemMessage(content=system_prompt), 
    HumanMessage(content=user_prompt)
]

response = llm_gemini.invoke(messages).content
```

```python
answer = textwrap.fill(str(response), width=50)
print(answer)
```

```python
from langchain_core.prompts import PromptTemplate
```

```python
prompt_template = """You have to behave like a movie star who is gay and is expert in teaching {topic} to beginners."""
prompt = PromptTemplate(
    input_variables = ["topic"], 
    template = prompt_template,
)

chain = prompt | llm_gemini
response = chain.invoke({"topic": "Explain me about how clouds are formed and how it rains through clouds in 10 lines."}).content
```

```python
answer = textwrap.fill(str(response), width = 50)
print(answer)
```

```python
# Langchain Expression language and Runnables.

from langchain_core.prompts import PromptTemplate

new_prompt = """You are a school teach who teaches kids from primary section and your speciality is to teach any concpet to anyone below 5 years old.
Teach this {concept} like you would according to your assigned role."""
prompt_template = PromptTemplate(
    input_variables = ["concept"], 
    template=new_prompt
)

```

```python
from langchain_core.output_parsers import StrOutputParser

output_parser = StrOutputParser()

chain = prompt_template | llm_gemini | output_parser

response = chain.invoke({"concept": "Nuclear Fission"})
print(response)
```

```python
# Using chain.stream() to get incremental output
print("Streaming response:\n")
for chunk in chain.stream({"concept": "Oppenheimer"}):
    print(chunk, end="", flush=True)
print("\n\n--- End of stream ---")
```

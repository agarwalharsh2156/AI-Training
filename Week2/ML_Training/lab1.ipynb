```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = [1.0, 2.0]
y = [300, 500]
print(x)
print(y)
```

```python
plt.scatter(x = x, y = y, marker= "s", c = "b")
plt.show()
```

```python
w = 100
b = 100
x_i = x
y_i = y
m = len(x)
print(w, b, x_i, y_i, m)
```

```python
y_hat = []
for i in range(m):
    f_xi = w * x_i[i] + b 
    y_hat.append(f_xi)
    print(f"Predicted price for {x_i[i]} is {f_xi}")
```

```python
print(y_hat)
```

```python
# Mean squared error
error = 0
for i in range(m):
    error += (y_hat[i] - y_i[i])**2
    print((y_hat[i] - y_i[i] ) ** 2)
print(error // (2*m))



# Cost function here:
# 1 / 2m * sum((y^hat_i - y^i)**2)
```

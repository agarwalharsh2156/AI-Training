### In order for the machine learning algorithm (here Linear Regression Model) to get the best set of weight and bias for accurate prediction, it is needed to reduce the cost function of the model. Gradient Descent is the widely used algorithm known for minimizing the cost function of a learning algorithm.

Gradient descent:
w = w - alpha * d J(w, b) / d w 
b = b - alpha * d J(w, b) / d w

Here J(w, b) is the cost function

If we talk about the Linear Regression Algorithm, J(w, b) = 1 / 2m * sum ([ f_wb(x^i) - y^i ] ** 2) where 1 >= i <= m
Here f(x^i) is nothing but the linear regression function f(x) = wx + b

```python
def compute_gradient(x, y, w, b):
    m = x.shape[0]
    w_d = 0
    b_d = 0
    
    y_hat = predict(x, w, b)
    for i in range(m):
        error = y_hat[i] - y[i]
        w_d += error * x[i]
        b_d += error
    
    w_d = (1/m) * w_d
    b_d = (1/m) * b_d
    return w_d, b_d


```

```python
def gradient_descent(x, y, w_in, b_in, alpha, num_iters):
    w = w_in
    b = b_in
    for i in range(num_iters):
        w_e, b_e = compute_gradient(x, y, w, b)
        w = w - (alpha * w_e)
        b = b - (alpha * b_e)

    return w, b

```

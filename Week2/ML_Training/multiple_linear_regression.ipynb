```python
# this has multiple features each assigned a different row for matrix dot product.
%%time
weights = np.array([0.1, 2, 3])
x = np.array([
    [100, 200, 300, 400],
    [10, 20, 30, 40],
    [1, 2, 3, 4]
])

bias = 10

y = np.dot(weights, x) + bias
print(y)
```

```python
import tensorflow as tf
```

```python
# Regression Neural Network.

model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation= "relu", input_shape = (12,)),
    tf.keras.layers.Dense(32, activation= "relu"),
    tf.keras.layers.Dense(1, activation="linear") # for regression don't mention any activation function or keep it linear.
])

model.compile(optimizer="adam", loss= "mse", metrics= ["mae"])
```

```python
# Binary classification
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation = "relu", input_shape = (12, )),
    tf.keras.layers.Dense(32, activation = "relu"),
    tf.keras.layers.Dense(1, activation = "sigmoid") # activation function to be sigmoid or tanh for binary classification.
])

model.compile(optimizer = "adam", loss="binary_crossentropy", metrics = ['accuracy']) # loss function for binary classification is binary cross entropy.
# metrics for both binary and multi class classification remains the same, i.e. accuracy.
```

```python
# Multiclass classification
model = tf.keras.Sequential([
    tf.keras.layers.Dense(64, activation = "relu", input_shape = (12, )),
    tf.keras.layers.Dense(32, activation = "relu"),
    tf.keras.layers.Dense(1, activation = "softmax") # activation function changes to softmax for multiclass classification.
])

model.compile(optimizer = "adam", loss="categorical_crossentropy", metrics = ['accuracy']) # loss function for multiclass classification is categorical crossentropy.
```

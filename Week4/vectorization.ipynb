```python
import numpy as np
import time
```

```python
# Power of vectorization.

a = np.random.rand(1000000)
b = np.random.rand(1000000)
tic = time.time()
vec_result = np.dot(a, b)
toc = time.time()
print(vec_result)
print('time taken: '+ str(1000 * (toc - tic)) + "ms")

# Multiplying using a for loop
c = 0
tic = time.time()
for i in range(1000000):
    c += a[i] * b[i]
toc = time.time()
print(c)
print("time taken", str(1000* (toc - tic)) + "ms")
```
